# Visualizing NYC Boroughs

Visualize the Boroughs of New York City with GeoPandas.

This example generates many images that are used in the documentation. See the [Geometric Manipulations](../docs/user_guide/geometric_manipulations.rst) example for more details.

First we'll import a dataset containing each borough in New York City. We'll use the ``geodatasets`` package to handle this quickly.

```python
import geodatasets
import matplotlib.pyplot as plt

import geopandas
```

```python
path_nybb = geodatasets.get_path("nybb")
boros = geopandas.read_file(path_nybb)
boros = boros.set_index("BoroCode")
boros
```

Next, we'll plot the raw data.

```python
ax = boros.plot()
plt.xticks(rotation=90);
```

We can easily retrieve the convex hull of each shape. This corresponds to the outer edge of the shapes.

```python
boros.geometry.convex_hull.plot(alpha=0.6)
plt.xticks(rotation=90);
```

We'll generate some random dots scattered throughout our data, and will use them to perform some set operations with our boroughs. We can use GeoPandas to perform unions, intersections, etc.

```python
N = 1000  # number of random points
R = 1500  # radius of buffer in feet

pts = boros.dissolve().sample_points(N, rng=1)
mp = pts.buffer(R).union_all()
boros_with_holes = boros.geometry.difference(mp)
boros_with_holes.plot()
plt.xticks(rotation=90);
```

Finally, we'll show the holes that were taken out of our boroughs.

```python
holes = boros.geometry.intersection(mp)
holes.plot()
plt.xticks(rotation=90);
```
