# Plotting with ipyleaflet
## What is ipyleaflet? 


ipyleaflet is a Jupyter widget for creating interactive maps based on the Leaflet JavaScript library. It allows GeoPandas objects to be displayed interactively in a notebook.

## ipyleaflet and GeoPandas `.explore()`
GeoPandas provides the `.explore()` method as a high-level convenience interface for creating interactive maps directly from a GeoDataFrame. It is a good default choice when you want to quickly inspect or share geospatial data with minimal setup.

`ipyleaflet` is lower-level and exposes the underlying Leaflet map as a Jupyter widget. This makes it useful when you need more control over the map object itself, such as adding custom layers, controls, or integrating the map with other Jupyter widgets.

## Load geometries

First, load a sample dataset as a GeoDataFrame. Web maps generally expect coordinates in latitude and longitude, so the data is converted to EPSG:4326 before displaying it.

```python
import geodatasets
from ipyleaflet import GeoData, LayersControl, Map

import geopandas

gdf = geopandas.read_file(geodatasets.get_path("geoda.chicago_commpop"))
gdf = gdf.to_crs(epsg=4326)

gdf.head()
```

## Create an ipyleaflet map

The map can be centered using the bounds of the GeoDataFrame.

```python
minx, miny, maxx, maxy = gdf.total_bounds
center = ((miny + maxy) / 2, (minx + maxx) / 2)

center
```

```python
m = Map(center=center, zoom=10)
```

## Add a GeoData layer

The `GeoData` layer accepts a GeoDataFrame directly and displays its geometries on the map.

```python
geo_data = GeoData(
    geo_dataframe=gdf,
    name="Chicago community areas",
    style={
        "color": "black",
        "fillColor": "#3366cc",
        "opacity": 1,
        "weight": 1,
        "fillOpacity": 0.5,
    },
)

m.add_layer(geo_data)
m.add_control(LayersControl())

m
```
